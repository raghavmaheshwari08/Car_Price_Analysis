###Meaning of Columns

Make- It indicates the brand of the car

Model- It is the model of the car (kind of a series)

Year- It is the year of manufacturing

Trim- The version of the car. Premium is better than base for the same model.

MSRP- The price for the customers

Invoice Price- The price for the dealer to pay to the manufacturer

Used/New Price- new price is the real time price of the car in the market whereas used price is for the cars having a resale value(used cars)

Rest other columns are describing the structural qualities and characteristics of the car, they are extremely important in describing the price.

###Converting into the right datatypes

```python
#MSRP, Invoice Price, Used/New Price should be converted into integer
#The dollar and also the commas need to be removed
df["MSRP"]=df["MSRP"].str.replace("$","")
df["MSRP"]=df["MSRP"].str.replace(",","")
```

```python
df["MSRP"]=df["MSRP"].astype(float)
```

```python
df["MSRP"]
```

```python
df["Invoice Price"]=df["Invoice Price"].str.replace("$","")
df["Invoice Price"]=df["Invoice Price"].str.replace(",","")
df["Invoice Price"]=df["Invoice Price"].astype(float)
```

```python
df["Used/New Price"]=df["Used/New Price"].str.replace("$","")
df["Used/New Price"]=df["Used/New Price"].str.replace(",","")
df["Used/New Price"]=df["Used/New Price"].astype(int)
```

```python
df["Used/New Price"]=df["Used/New Price"].astype(float)
```

```python
df["Horsepower"].value_counts()
```

```python
#Dividing the horsepower column into two separate columns out of which we need the first column
#The first column will have the peak horspower value at a certain rpm value. For the price, basically the peak hp is sufficient
```

```python
import numpy as np

df["Horsepower"]=df["Horsepower"].str.split(" ").str[0]
df["Horsepower"]

df["Horsepower"]=df["Horsepower"].astype(float)
```

```python
df["Torque"]=df["Torque"].str.split(" ").str[0]
df["Torque"]

df["Torque"]=df["Torque"].astype(float)
```

```python
df["Highway Fuel Economy"]=df["Highway Fuel Economy"].str.split(" ").str[0]
df["Highway Fuel Economy"]

df["Highway Fuel Economy"]=df["Highway Fuel Economy"].astype(float)
```
